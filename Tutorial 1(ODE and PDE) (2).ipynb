# 3) Numerical approaches
## 1) Rectangular method

```python
def rectangular_right(f1,a,b,N):
    h,X = maillage(a,b,N)
    I_right=0
    for i in range(1,N+1):
        I_right = I_right + f1(X[i])
    I_right = h*I_right

    return I_right
I_right = rectangular_right(f1,a,b,N)

print("approximate value for I: ", I_right)  
print("error equal: ", abs(Iexact-I_right))
```
