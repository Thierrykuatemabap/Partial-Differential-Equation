```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
D = 0.5
r = 1.0
dx = 0.01
dt = 0.0001
x = np.arange(0 , 2 , dx)

# Initial Gaussian perturbation
u = np.exp( -50*( x -0.5) **2)

def step(u , D , r , dx , dt):
    u_new = u.copy()
    u_new [1: -1] = (u [1: -1]+ D * dt / dx **2*( u [2:] - 2* u [1: -1] + u [: -2])
                    + r * dt * u[1: -1]*(1 - u [1: -1]) )
    return u_new

# Time evolution
for t in range (1000) :
    u = step (u , D , r , dx , dt)

plt.plot(x , u )
plt.title("Fisher -KPP Reaction - Diffusion at t=1")
plt.xlabel("x")
plt.ylabel("u(x,t)")
plt.grid( True )
plt.show()
```
