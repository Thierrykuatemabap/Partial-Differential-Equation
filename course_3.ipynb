```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation ,PillowWriter

# Parameters
c = 2.0
x = np.linspace(-10 , 10, 500)
t_max = 5
frames = 150

 # Solution
def u(x, t):
  return 0.5 * (np.exp(-(x - c * t )**2) + np.exp (-(x+ c * t)**2) )


 # Figure setup
fig , ax = plt.subplots()
line , = ax.plot(x , u(x , 0) , color ='blue', lw =2)

ax.set_xlim(-10 , 10)
ax.set_ylim(0 , 1)
ax.set_xlabel("x")
ax.set_ylabel("u(x,t)")
ax.set_title("Gaussian Pulses( Wave Equation )")

 # Update function
def update(frame):
  t = t_max * frame / frames
  y = u(x , t)
  line.set_ydata( y )
  ax.set_title(f" Gaussian Pulses at t = {t:.2f}")
  return line ,

 # Create animation
ani = FuncAnimation(fig , update, frames=frames ,interval=50)


 # Save as GIF
ani.save("gaussian_pulses2.gif", writer=PillowWriter( fps=20))

plt.close()
```

```python
import numpy as np
import matplotlib . pyplot as plt
from matplotlib . animation import FuncAnimation ,PillowWriter


# Parameters

L = 10.0 # String length
c = 1.0 # Wave speed
t_max = 20.0 # Maximum time
frames = 200 # Number of frames in GIF
x = np.linspace(0 , L , 500)

# Mode numbers to illustrate
mode_numbers = [1 , 2 , 3]

# Function to compute resonant solution
def u_resonant(x , t , n ):
    k_n = n * np . pi / L
    return ( t / (2 * c * k_n ) ) * np . sin ( k_n *x )


# 1. Static plots for different modes

plt.figure ( figsize =(8 ,5) )

for n in mode_numbers:
    plt.plot (x , u_resonant (x , t_max , n ) ,label = f"n={n}")
    plt.xlabel("x")
    plt.ylabel("u(x,t)")
    plt.title(f'Resonant string profiles at t={t_max}')
    plt.legend()
    plt.grid( True )
    plt.show()

# 2. Contour map of u(x,t) over x and t for mode

n=1

t = np.linspace (0 , t_max , 300)
X , T = np.meshgrid (x , t )
U = u_resonant(X , T , n =1)

plt.figure( figsize =(8 ,5) )
plt.contourf(X , T , U , 50 , cmap ='viridis')
plt.colorbar( label ='u(x,t)')
plt.xlabel("x")
plt.ylabel("t")
plt.title("Contour of u(x,t) for mode n=1 (resonance )")
plt.show()

# 3. GIF animation for mode n=1

fig , ax = plt.subplots()
line , = ax.plot(x , u_resonant (x , 0 , n =1) , lw =2)
ax.set_xlim(0 , L )
ax.set_ylim( - np .max( U ) *1.1 , np .max( U ) *1.1)
ax.set_xlabel("x")
ax.set_ylabel("u(x,t)")
ax.set_title("Resonant mode animation n=1")

def update( frame ):
    t_frame = t_max * frame / frames
    y = u_resonant(x , t_frame , n =1)
    line.set_ydata( y )
    ax.set_title(f"Resonant mode n=1 , t={t_frame:.2f}")
    return line ,

ani = FuncAnimation ( fig , update , frames = frames ,interval =50)

ani.save("resonant_mode1.gif", writer =PillowWriter ( fps =20) )

plt.close()
```
